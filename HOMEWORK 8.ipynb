```python
class Shape:
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
        
    def area(self):
        return self.w * self.h

class Circle(Shape):
    def __init__(self, r):
        self.r = r
        
    def area(self):
        return 3.14 * (self.r ** 2)


shape_type = input().strip().upper()
params = input().strip().split()


shape = None

if shape_type == "RECTANGLE":
    w = float(params[0])
    h = float(params[1])
    shape = Rectangle(w, h)
elif shape_type == "CIRCLE":
    r = float(params[0])
    shape = Circle(r)


if shape is not None:
    print(f"{shape.area():.2f}")
```

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
        
    def info(self):
        return f"{self.brand}-{self.speed}"

class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats = seats
        
    def info(self):
        return f"{self.brand}-{self.speed}-{self.seats}"

obj_type = input().strip().lower()
data = input().strip().split()

if obj_type == "vehicle":
    brand = data[0]
    speed = data[1]
    obj = Vehicle(brand, speed)
elif obj_type == "car":
    brand = data[0]
    speed = data[1]
    seats = data[2]
    obj = Car(brand, speed, seats)

print(obj.info())
```

```python
class Polygon:
    def perimeter(self):
        pass

class Triangle(Polygon):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
        
    def perimeter(self):
        return self.a + self.b + self.c

data = input().strip().split()
a = int(data[0])
b = int(data[1])
c = int(data[2])

triangle = Triangle(a, b, c)
print(triangle.perimeter())
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def describe(self):
        return f"{self.name} {self.age}"

class Student(Person):
    def __init__(self, name, age, school):
        super().__init__(name, age)
        self.school = school
        
    def describe(self):
        return f"{self.name} {self.age} {self.school}"

obj_type = input().strip().lower()
data = input().strip().split()

if obj_type == "person":
    obj = Person(data[0], data[1])
elif obj_type == "student":
    obj = Student(data[0], data[1], data[2])

print(obj.describe())
```

```python
class Account:
    def __init__(self, balance):
        self.balance = balance
        
    def deposit(self, x):
        self.balance += x
        
    def get_balance(self):
        return self.balance

class SavingAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate
        
    def add_interest(self):
        self.balance += self.balance * self.rate

data = input().strip().split()
balance = float(data[0])
rate = float(data[1])

account = SavingAccount(balance, rate)
account.add_interest()

print(f"{account.get_balance():.2f}")
```

```python
class Printer:
    def print_value(self, x):
        pass

class BinaryPrinter(Printer):
    def print_value(self, x):
        return bin(x)[2:]

x = int(input().strip())

printer = BinaryPrinter()
print(printer.print_value(x))
```

```python
class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
        
    def trace(self):
        return self.a + self.d
        
    def determinant(self):
        return (self.a * self.d) - (self.b * self.c)

data = input().strip().split()
a = int(data[0])
b = int(data[1])
c = int(data[2])
d = int(data[3])

matrix = Matrix2x2(a, b, c, d)
print(matrix.trace(), matrix.determinant())
```

```python
class Polynomial:
    def __init__(self, a, b):
        self.a = a
        self.b = b
        
    def evaluate(self, x):
        return self.a * x + self.b

data = input().strip().split()
a = float(data[0])
b = float(data[1])
x = float(data[2])

poly = Polynomial(a, b)
result = poly.evaluate(x)

if result.is_integer():
    print(int(result))
else:
    print(result)
```

```python
class Polynomial2:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
        
    def evaluate(self, x):
        return self.a * (x ** 2) + self.b * x + self.c

data = input().strip().split()
a = float(data[0])
b = float(data[1])
c = float(data[2])
x = float(data[3])

poly2 = Polynomial2(a, b, c)
result = poly2.evaluate(x)

if result.is_integer():
    print(int(result))
else:
    print(result)
```

```python
class PolynomialSum:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
        
    def sum_str(self):
        k = self.a + self.c
        m = self.b + self.d
        
        if m >= 0:
            return f"{k}x + {m}"
        else:
            return f"{k}x - {abs(m)}"

data = input().strip().split()
a = int(data[0])
b = int(data[1])
c = int(data[2])
d = int(data[3])

poly_sum = PolynomialSum(a, b, c, d)
print(poly_sum.sum_str())
```
