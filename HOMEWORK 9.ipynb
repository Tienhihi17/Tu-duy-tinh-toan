```python
def generate(s, zeros, ones):
    if zeros == 0 and ones == 0:
        print(s)
        return
    if zeros > 0:
        generate(s + "0", zeros - 1, ones)
    if ones > 0:
        generate(s + "1", zeros, ones - 1)

data = input().strip().split()
n = int(data[0])
m = int(data[1])

generate("", n, m)
```

```python
arr = [int(x) for x in input().strip().split()]

if not arr:
    result = []
else:
    n = len(arr)
    dp = [1] * n
    trace = [-1] * n

    for i in range(1, n):
        for j in range(i):
            if arr[i] > arr[j] and dp[i] < dp[j] + 1:
                dp[i] = dp[j] + 1
                trace[i] = j

    max_len = max(dp)
    last_idx = dp.index(max_len)

    result = []
    while last_idx != -1:
        result.append(arr[last_idx])
        last_idx = trace[last_idx]
    result.reverse()

print(*(result))
```

```python
s = input()

words = []
current_word = ""

for char in s:
    if char != " " and char != "\t" and char != "\n":
        current_word += char
    else:
        if current_word != "":
            words.append(current_word)
            current_word = ""
if current_word != "":
    words.append(current_word)

result = ""
for i in range(len(words)):
    word = words[i]
    capitalized_word = ""
    
    first_char = word[0]
    if "a" <= first_char <= "z":
        capitalized_word += chr(ord(first_char) - 32)
    else:
        capitalized_word += first_char
        
    for j in range(1, len(word)):
        char = word[j]
        if "A" <= char <= "Z":
            capitalized_word += chr(ord(char) + 32)
        else:
            capitalized_word += char
            
    if i == 0:
        result += capitalized_word
    else:
        result += " " + capitalized_word

print(result)
```

```python
def partition(arr, low, high):
    pivot = arr[high]
    i = low
    for j in range(low, high):
        if arr[j] >= pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1
    arr[i], arr[high] = arr[high], arr[i]
    return i

def quickselect(arr, low, high, k):
    if low <= high:
        pivot_index = partition(arr, low, high)
        if pivot_index == k - 1:
            return arr[pivot_index]
        elif pivot_index > k - 1:
            return quickselect(arr, low, pivot_index - 1, k)
        else:
            return quickselect(arr, pivot_index + 1, high, k)
    return -1

arr = [int(x) for x in input().strip().split()]
k = int(input().strip())

print(quickselect(arr, 0, len(arr) - 1, k))
```

```python
num_str = input().strip()
k = int(input().strip())

to_remove = len(num_str) - k
stack = []

for digit in num_str:
    while to_remove > 0 and stack and stack[-1] < digit:
        stack.pop()
        to_remove -= 1
    stack.append(digit)

result = "".join(stack[:k])
print(result)
```

```python
s = input().strip()

if not s:
    print("")
else:
    result = ""
    count = 1
    
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            result += s[i-1] + str(count)
            count = 1
            
    result += s[-1] + str(count)
    print(result)
```

```python
password = input().strip()

has_lower = False
has_upper = False
has_digit = False
has_special = False
special_chars = "!@#$"

for char in password:
    if "a" <= char <= "z":
        has_lower = True
    elif "A" <= char <= "Z":
        has_upper = True
    elif "0" <= char <= "9":
        has_digit = True
    elif char in special_chars:
        has_special = True

if len(password) >= 8 and has_lower and has_upper and has_digit and has_special:
    print("STRONG")
else:
    print("WEAK")
```
