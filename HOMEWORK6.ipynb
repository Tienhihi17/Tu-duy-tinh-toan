```python
def W6A18():
    n, m, k = map(int, input().split())
    arr = [list(map(int, input().split())) for _ in range(n)]
    print(f"Tong cot {k}:", sum(arr[i][k] for i in range(n)))
    print(f"Tong hang {k}:", sum(arr[k][j] for j in range(m)))
W6A18()
```

```python
def tim_vi_tri_max(lst):
    max_idx = 0
    for i in range(1, len(lst)):
        if lst[i] > lst[max_idx]:
            max_idx = i
    return max_idx
```

```python
 
def W6A19():
    lst = list(map(int, input().split()))
    print(tim_vi_tri_max(lst))
W6A19()
 
def tim_vi_tri_k(lst, k):
    for i, v in enumerate(lst):
        if v == k:
            return i
    return -1
```

```python
def W6A20():
    lst = list(map(int, input().split()))
    k = int(input())
    print(tim_vi_tri_k(lst, k))
W6A20()
```

```python
def W6A24():
    lst = list(map(int, input().split()))
    choice = input("Day 0 ve dau (d) hay cuoi (c)? ").strip().lower()
    if choice == 'd':
        print([0] * lst.count(0) + [v for v in lst if v != 0])
    else:
        print([v for v in lst if v != 0] + [0] * lst.count(0))
W6A24()
```

```python
def W6A27():
    lst = list(map(int, input().split()))
    x = int(input())
    positions = [i for i, v in enumerate(lst) if v < x]
    print(positions if positions else -1)
W6A27()
 
```

```python
def W6A28():
    lst = list(map(int, input().split()))
    tang = all(lst[i] < lst[i+1] for i in range(len(lst)-1))
    giam = all(lst[i] > lst[i+1] for i in range(len(lst)-1))
    print("Tang thuc su" if tang else "Giam thuc su" if giam else "Vo danh")
W6A28()
```
