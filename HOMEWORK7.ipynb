Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu 
mỗi phần tử x trong mảng xuất hiện đúng x lần. 
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa 
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp. 

```python
def make_good_array(arr):
    counts = {}
    for x in arr:
        counts[x] = counts.get(x, 0) + 1
    
    y = 0
    for x, c in counts.items():
        if c >= x:
            y += (c - x)
        else:
            y += c
    return y

print(make_good_array([3,3,3,1]))
```

Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ 
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore. 
Nếu không có phần tử nào thỏa mãn thì không in gì 

```python
def find_majority(arr):
    candidate = None
    count = 0
    for x in arr:
        if count == 0:
            candidate = x
            count = 1
        elif x == candidate:
            count += 1
        else:
            count -= 1
    
    # Kiểm tra lại xem candidate có thực sự chiếm > n/2 không
    if arr.count(candidate) > len(arr) // 2:
        print(candidate)
print(find_majority([3, 3, 3, 1]))
```

Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng 
dần, lẻ giảm dần; ;  

```python
def conditional_sort(arr):
    chan = sorted([x for x in arr if x % 2 == 0])
    le = sorted([x for x in arr if x % 2 != 0], reverse=True)
    return chan + le
print(conditional_sort([5, 2, 3, 8, 1, 4]))
```

 
Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không 
dùng bất cứ thuật toán sắp xếp nào 

```python
def merge_sorted(A, B):
    i, j = 0, 0
    res = []
    while i < len(A) and j < len(B):
        if A[i] < B[j]:
            res.append(A[i]); i += 1
        else:
            res.append(B[j]); j += 1
    return res + A[i:] + B[j:] 
print(merge_sorted([1, 3, 5], [2, 4, 6]))
```

Cho N là ngày, T là tháng năm 2020, Tính xem đó là ngày thứ bao nhiêu 
của năm đó nếu đó là ngảy hợp lệ 

```python
def day_of_year(d, t):
    days_in_month = [0, 31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    if 1 <= t <= 12 and 1 <= d <= days_in_month[t]:
        return sum(days_in_month[:t]) + d
    return "Ngày không hợp lệ"
print(day_of_year(15, 3))
```

Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất. 
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất 

```python
def most_frequent_name(names):
    counts = {}
    for name in names:
        counts[name] = counts.get(name, 0) + 1
    
    # Sắp xếp: -x[1] là số lần giảm dần, x[0] là tên tăng dần (từ điển)
    res = sorted(counts.items(), key=lambda x: (-x[1], x[0]))
    return res[0][0]
print(most_frequent_name(["Alice", "Bob", "Alice", "Charlie", "Bob", "Bob"]))
```

Đếm số lượng tên trong xâu kí tự. Tên được định nghĩa là xâu có độ dài 
>=2 và có chữ cái đầu viết hoa, các chữ cái khác viết thường, chỉ có kí 
tự cái. 

```python
import re
def count_valid_names(text):
    # Regex: Chữ cái Hoa + các chữ cái thường (ít nhất 1 chữ thường để tổng >= 2)
    pattern = r'\b[A-Z][a-z]+\b'
    matches = re.findall(pattern, text)
    return len(matches)

print(count_valid_names("Alice and Bob are friends. Charlie is also a friend."))
```

 Cho xâu kí tự, xóa các từ giống nhau liền kề, chỉ giữ lại từ đầu tiên 

```python
def remove_adjacent_duplicates(s):
    words = s.split()
    if not words: return ""
    
    res = [words[0]]
    for i in range(1, len(words)):
        if words[i] != words[i-1]:
            res.append(words[i])
    return " ".join(res)

print(remove_adjacent_duplicates("hello hello world world world"))
```

Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi 
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số 

```python
def largest_k_digit_number(n_str, k):
    to_remove = len(n_str) - k
    stack = []
    for digit in n_str:
        while to_remove > 0 and stack and stack[-1] < digit:
            stack.pop()
            to_remove -= 1
        stack.append(digit)
    
    return "".join(stack[:k])

print(largest_k_digit_number("1432219", 3))
```

Tách các số có trong một đoạn văn bản, tính tổng và tính trung bình các 
số đó.  

```python
import re

def process_numbers_in_text(text):
    nums = list(map(int, re.findall(r'\d+', text)))
    if not nums: return 0, 0
    
    tong = sum(nums)
    trung_binh = tong / len(nums)
    return tong, trung_binh
print(process_numbers_in_text("The numbers are 10, 20 and 30."))
```

Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ 
hay không 

```python
def is_valid_parentheses(s):
    stack = []
    mapping = {")": "(", "}": "{", "]": "["}
    for char in s:
        if char in mapping.values():
            stack.append(char)
        elif char in mapping:
            if not stack or stack.pop() != mapping[char]:
                return False
    return len(stack) == 0
print(is_valid_parentheses("()[]{}"))
```

Bà ̀ i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên 
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ 
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua) 

```python
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0: return False
    return True

def longest_composite_sequence(N):
    max_seq = []
    current_seq = []
    
    for i in range(2, N + 1):
        if not is_prime(i):
            current_seq.append(i)
        else:
            if len(current_seq) > len(max_seq):
                max_seq = current_seq
            current_seq = []
    
    # Check lần cuối
    if len(current_seq) > len(max_seq):
        max_seq = current_seq
    return max_seq
print(longest_composite_sequence(30))
```

Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập 
vào từ bàn phím  

```python
def longest_increasing_subsequence(arr):
    n = len(arr)
    if n == 0: return []
    # f[i] lưu độ dài dãy con tăng dài nhất kết thúc tại i
    f = [1] * n
    # trace[i] để truy vết lại dãy
    trace = [-1] * n
    
    for i in range(n):
        for j in range(i):
            if arr[i] > arr[j] and f[i] < f[j] + 1:
                f[i] = f[j] + 1
                trace[i] = j
                
    max_len = max(f)
    idx = f.index(max_len)
    
    # Truy vết ngược để lấy dãy
    res = []
    while idx != -1:
        res.append(arr[idx])
        idx = trace[idx]
    return res[::-1]
print(longest_increasing_subsequence([10, 20, 10, 30, 20, 50]))
```
